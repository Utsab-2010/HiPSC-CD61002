```python
def steepest_desc(A,b,tol=1e-6,maxiter=1000):
    x = np.zeros(len(b))
    r = b - A @ x
    p = A @ r
    for epoch in range(maxiter):
        # print(epoch, np.linalg.norm(r))
        if np.linalg.norm(r) < tol:
            print("Steepest Descent converged in", epoch, "iterations")
            break
        alpha = (r @ r) / (p @ r)
        x = x + alpha * r
        r = b - A @ x
        p = A @ r
    return x
```

```python
def min_res(A,b,tol=1e-6,maxiter=1000):
    x = np.zeros(len(b))
    r = b - A@x
    p = A @ r
    for epoch in range(maxiter):
        if np.linalg.norm(r) < tol:
            print("Min Residual converged in", epoch, "iterations")
            break
        alpha = (p@r)/(p@p)
        x = x + alpha*r
        r = r - alpha*p
        p = A @ r
    return x
```

```python
x = min_res(A,B)
```

```python
def conj_grad(A,b,tol=1e-6,maxiter=1000):
    x = np.zeros(len(b))
    r = b - A@x
    p = r
    for epoch in range(maxiter):
        if np.linalg.norm(r) < tol:
            print("Conjugate Gradient converged in", epoch, "iterations")
            break
        Ap = A @ p
        alpha = (r@r)/(p@Ap)
        x = x + alpha*p
        r_new = r - alpha*Ap
        beta = (r_new @ r_new)/(r @ r)
        p = r_new + beta*p
        r = r_new
    return x
```
